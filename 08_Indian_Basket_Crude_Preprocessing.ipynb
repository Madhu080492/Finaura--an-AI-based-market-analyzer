```python
raw = pd.read_excel(input_path, header=None)

print("Shape:", raw.shape)
raw.head(20)
```

```python
for i in range(8, 16):
    print("ROW", i)
    print(raw.iloc[i].tolist())
    print()
```

```python
df = raw.iloc[12:].copy()

df = df.iloc[:, :15]

df.columns = [
    "financial_year",
    "apr",
    "may",
    "jun",
    "jul",
    "aug",
    "sep",
    "oct",
    "nov",
    "dec",
    "jan",
    "feb",
    "mar",
    "average",
    "ratio"
]

df = df.reset_index(drop=True)

df.head()
```

```python
print("Shape:", df.shape)
print(df.columns.tolist())
```

```python
df = df.dropna(
    subset=["financial_year"]
).reset_index(drop=True)

print("Shape after removing empty rows:", df.shape)
```

```python
print(df["financial_year"].tolist())
```

```python
month_columns = [
    "apr", "may", "jun", "jul",
    "aug", "sep", "oct", "nov",
    "dec", "jan", "feb", "mar"
]

for col in month_columns:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

df[month_columns].head()
```

```python
df["average"] = pd.to_numeric(
    df["average"],
    errors="coerce"
)

df[["financial_year", "average"]].head()
```

```python
print("Missing values:")
print(df.isnull().sum())
```

```python
print(
    "Duplicate financial years:",
    df["financial_year"].duplicated().sum()
)
```

```python
df = df.drop_duplicates(
    subset=["financial_year"]
).reset_index(drop=True)

print("Shape:", df.shape)
```

```python
long_df = df.melt(
    id_vars=["financial_year"],
    value_vars=month_columns,
    var_name="month",
    value_name="crude_oil_price"
)

long_df.head(15)
```

```python
month_mapping = {
    "apr": 4,
    "may": 5,
    "jun": 6,
    "jul": 7,
    "aug": 8,
    "sep": 9,
    "oct": 10,
    "nov": 11,
    "dec": 12,
    "jan": 1,
    "feb": 2,
    "mar": 3
}

long_df["month_number"] = long_df["month"].map(
    month_mapping
)

long_df.head()
```

```python
long_df["financial_year_start"] = (
    long_df["financial_year"]
    .astype(str)
    .str.extract(r"(\d{4})")[0]
)

long_df = long_df.dropna(
    subset=["financial_year_start"]
).copy()

long_df["financial_year_start"] = (
    long_df["financial_year_start"].astype(int)
)

long_df.head()
```

```python
long_df["year"] = np.where(
    long_df["month_number"] >= 4,
    long_df["financial_year_start"],
    long_df["financial_year_start"] + 1
)

long_df.head(15)
```

```python
long_df["date"] = pd.to_datetime(
    dict(
        year=long_df["year"].astype(int),
        month=long_df["month_number"].astype(int),
        day=1
    ),
    errors="coerce"
)

long_df.head(15)
```

```python
crude = long_df[
    [
        "date",
        "crude_oil_price"
    ]
].copy()

crude = crude.sort_values(
    "date"
).reset_index(drop=True)

crude.head(15)
```

```python
print("Start Date:", crude["date"].min())
print("End Date:", crude["date"].max())
print("Rows:", len(crude))
```

```python
print(
    "Missing crude oil prices:",
    crude["crude_oil_price"].isnull().sum()
)
```

```python
print(
    "Non-positive prices:",
    (crude["crude_oil_price"] <= 0).sum()
)
```

```python
crude = crude[
    crude["crude_oil_price"] > 0
].copy()

crude = crude.reset_index(drop=True)

print("Shape:", crude.shape)
```

```python
crude["previous_crude_price"] = (
    crude["crude_oil_price"].shift(1)
)

crude.head()
```

```python
crude["crude_price_change"] = (
    crude["crude_oil_price"]
    - crude["previous_crude_price"]
)

crude.head()
```

```python
crude["crude_return"] = (
    crude["crude_oil_price"].pct_change()
)

crude.head()
```

```python
crude["log_return"] = np.log(
    crude["crude_oil_price"]
    / crude["previous_crude_price"]
)

crude.head()
```

```python
crude["crude_ma_3"] = (
    crude["crude_oil_price"]
    .rolling(3)
    .mean()
)

crude["crude_ma_6"] = (
    crude["crude_oil_price"]
    .rolling(6)
    .mean()
)

crude["crude_ma_12"] = (
    crude["crude_oil_price"]
    .rolling(12)
    .mean()
)

crude.tail()
```

```python
crude["volatility_3m"] = (
    crude["crude_return"]
    .rolling(3)
    .std()
)

crude["volatility_6m"] = (
    crude["crude_return"]
    .rolling(6)
    .std()
)

crude["volatility_12m"] = (
    crude["crude_return"]
    .rolling(12)
    .std()
)

crude.tail()
```

```python
crude["momentum_3m"] = (
    crude["crude_oil_price"]
    / crude["crude_oil_price"].shift(3)
    - 1
)

crude["momentum_6m"] = (
    crude["crude_oil_price"]
    / crude["crude_oil_price"].shift(6)
    - 1
)

crude.tail()
```

```python
crude["price_vs_ma_3"] = (
    crude["crude_oil_price"]
    / crude["crude_ma_3"]
    - 1
)

crude["price_vs_ma_6"] = (
    crude["crude_oil_price"]
    / crude["crude_ma_6"]
    - 1
)

crude["price_vs_ma_12"] = (
    crude["crude_oil_price"]
    / crude["crude_ma_12"]
    - 1
)

crude.tail()
```

```python
crude["target_next_month_crude_price"] = (
    crude["crude_oil_price"].shift(-1)
)

crude.tail()
```

```python
crude["target_next_month_change"] = (
    crude["target_next_month_crude_price"]
    - crude["crude_oil_price"]
)

crude.tail()
```

```python
feature_columns = [
    "previous_crude_price",
    "crude_price_change",
    "crude_return",
    "log_return",
    "crude_ma_3",
    "crude_ma_6",
    "crude_ma_12",
    "volatility_3m",
    "volatility_6m",
    "volatility_12m",
    "momentum_3m",
    "momentum_6m",
    "price_vs_ma_3",
    "price_vs_ma_6",
    "price_vs_ma_12"
]

target_columns = [
    "target_next_month_crude_price",
    "target_next_month_change"
]

print("Feature columns:")
print(feature_columns)

print("\nTarget columns:")
print(target_columns)
```

```python
print(
    crude[target_columns].isnull().sum()
)
```

```python
crude = crude.dropna().reset_index(drop=True)

print("Final shape:", crude.shape)
```

```python
print("FINAL DATASET VALIDATION")
print("------------------------")

print("Rows:", len(crude))
print("Columns:", len(crude.columns))

print(
    "Missing values:",
    crude.isnull().sum().sum()
)

print(
    "Duplicate rows:",
    crude.duplicated().sum()
)

print(
    "Date range:",
    crude["date"].min(),
    "to",
    crude["date"].max()
)
```
