```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
input_path = "../data/processed/nifty50_processed.csv"

df = pd.read_csv(input_path)

df["date"] = pd.to_datetime(df["date"])

print("Shape:", df.shape)
```

```python
df.info()
```

```python
df.describe()
```

```python
print(df.isnull().sum())
```

```python
print("Start Date:", df["date"].min())
print("End Date:", df["date"].max())
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    df["date"],
    df["close"]
)

plt.xlabel("Date")
plt.ylabel("NIFTY 50 Close Price")
plt.title("NIFTY 50 Price Trend")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    df["date"],
    df["daily_return"]
)

plt.xlabel("Date")
plt.ylabel("Daily Return")
plt.title("NIFTY 50 Daily Returns")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["daily_return"].dropna(),
    bins=50
)

plt.xlabel("Daily Return")
plt.ylabel("Frequency")
plt.title("Distribution of NIFTY 50 Daily Returns")

plt.tight_layout()
plt.show()
```

```python
print("Available columns:")
print(df.columns.tolist())
```

```python
volatility_columns = [
    col for col in df.columns
    if "volatility" in col.lower()
]

print("Volatility columns:")
print(volatility_columns)
```

```python
if "volatility_20d" in df.columns:
    volatility_col = "volatility_20d"
elif "volatility_20" in df.columns:
    volatility_col = "volatility_20"
else:
    volatility_col = volatility_columns[-1]

print("Selected volatility column:", volatility_col)
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    df["date"],
    df[volatility_col]
)

plt.xlabel("Date")
plt.ylabel("Annualized Volatility")
plt.title("NIFTY 50 Rolling Volatility")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df[volatility_columns].describe()
```

```python
df.nlargest(
    10,
    volatility_col
)[
    [
        "date",
        "close",
        volatility_col
    ]
]
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["daily_return"].dropna(),
    bins=50
)

plt.xlabel("Daily Return")
plt.ylabel("Frequency")
plt.title("Distribution of NIFTY 50 Daily Returns")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["log_return"].dropna(),
    bins=50
)

plt.xlabel("Log Return")
plt.ylabel("Frequency")
plt.title("Distribution of NIFTY 50 Log Returns")

plt.tight_layout()
plt.show()
```

```python
ma_columns = [
    col for col in df.columns
    if col.startswith("ma_")
]

print("Moving Average columns:")
print(ma_columns)
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    df["date"],
    df["close"],
    label="Close Price"
)

for col in ma_columns:
    if col in ["ma_20", "ma_50"]:
        plt.plot(
            df["date"],
            df[col],
            label=col
        )

plt.xlabel("Date")
plt.ylabel("Price")
plt.title("NIFTY 50 Price and Moving Averages")
plt.legend()

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
print("Target Variable Statistics")
print("--------------------------")

print(
    df["target_5d_volatility"].describe()
)
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["target_5d_volatility"].dropna(),
    bins=40
)

plt.xlabel("Next 5-Day Realized Volatility")
plt.ylabel("Frequency")
plt.title("Distribution of Target Variable")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 5))

plt.scatter(
    df[volatility_col],
    df["target_5d_volatility"],
    alpha=0.5
)

plt.xlabel("Current Rolling Volatility")
plt.ylabel("Next 5-Day Volatility")
plt.title("Current Volatility vs Future Volatility")

plt.tight_layout()
plt.show()
```

```python
correlation_columns = [
    "daily_return",
    "log_return",
    "intraday_return",
    "price_range",
    "momentum_5",
    "momentum_20",
    "drawdown"
]

correlation_columns = [
    col for col in correlation_columns
    if col in df.columns
]

correlation_columns += volatility_columns

print("Columns used for correlation:")
print(correlation_columns)
```

```python
correlation_matrix = df[
    correlation_columns
].corr()

correlation_matrix
```

```python
plt.figure(figsize=(10, 8))

plt.imshow(
    correlation_matrix,
    aspect="auto"
)

plt.colorbar()

plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    rotation=90
)

plt.yticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns
)

plt.title("NIFTY 50 Feature Correlation Matrix")

plt.tight_layout()
plt.show()
```

```python
print("Target column check:")
print(df["target_5d_volatility"].shape)

print("\nTarget column names matching:")
print([
    col for col in df.columns
    if col == "target_5d_volatility"
])

print("\nCorrelation columns:")
print(correlation_columns)
```

```python
corr = df.corr(numeric_only=True)

print("Correlation matrix shape:", corr.shape)

print("\nTarget exists in correlation matrix:")
print("target_5d_volatility" in corr.columns)

print("\nTarget correlation:")
print(corr["target_5d_volatility"])
```

```python
corr = df.corr(numeric_only=True)

target_correlations = corr[
    "target_5d_volatility"
].sort_values(ascending=False)

print(target_correlations)
```

```python
strongest_features = (
    target_correlations
    .drop("target_5d_volatility")
    .sort_values(
        key=abs,
        ascending=False
    )
)

print("Features ranked by absolute correlation with target:")
print(strongest_features)
```

```python
print("NIFTY 50 EDA SUMMARY")
print("====================")

print("Dataset shape:", df.shape)

print(
    "Date range:",
    df["date"].min(),
    "to",
    df["date"].max()
)

print(
    "Average daily return:",
    df["daily_return"].mean()
)

print(
    "Average future 5-day volatility:",
    df["target_5d_volatility"].mean()
)

print(
    "Maximum rolling volatility:",
    df[volatility_col].max()
)

print(
    "Minimum rolling volatility:",
    df[volatility_col].min()
)
```

```python
eda_summary = pd.DataFrame({
    "Metric": [
        "Rows",
        "Columns",
        "Start Date",
        "End Date",
        "Average Daily Return",
        "Average Future 5-Day Volatility",
        "Maximum Rolling Volatility"
    ],
    "Value": [
        len(df),
        len(df.columns),
        df["date"].min(),
        df["date"].max(),
        df["daily_return"].mean(),
        df["target_5d_volatility"].mean(),
        df[volatility_col].max()
    ]
})

eda_summary
```

```python
volatility_columns = [
    col for col in df.columns
    if "volatility" in col.lower()
]

print("Volatility columns:", volatility_columns)

volatility_col = volatility_columns[-1]

summary = {
    "rows": len(df),
    "columns": len(df.columns),
    "date_start": df["date"].min(),
    "date_end": df["date"].max(),
    "missing_values": df.isnull().sum().sum(),
    "duplicate_rows": df.duplicated().sum(),
    "mean_close": df["close"].mean(),
    "min_close": df["close"].min(),
    "max_close": df["close"].max(),
    "mean_daily_return": df["daily_return"].mean(),
    "return_std": df["daily_return"].std(),
    "min_daily_return": df["daily_return"].min(),
    "max_daily_return": df["daily_return"].max(),
    "mean_volatility": df[volatility_col].mean(),
    "max_volatility": df[volatility_col].max(),
    "mean_future_volatility": df["target_5d_volatility"].mean(),
    "max_future_volatility": df["target_5d_volatility"].max()
}

summary_df = pd.DataFrame([summary])

summary_df
```

```python
output_path = "../data/processed/nifty50_eda_summary.csv"

summary_df.to_csv(
    output_path,
    index=False
)

print("NIFTY 50 EDA summary saved successfully!")
print("Saved at:", output_path)
```
